```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
sns.countplot(data=penguins, x="species")
plt.title("Number of penguins per species")
plt.show()
```

```python
penguins["species"].value_counts()
```

```python
sns.histplot(data=penguins, x="body_mass_g", kde=True)
plt.title("Body mass distribution")
plt.show()
```

```python
sns.histplot(data=penguins, x="body_mass_g", bins=10)
plt.title("10 bins")
plt.show()

sns.histplot(data=penguins, x="body_mass_g", bins=40)
plt.title("40 bins")
plt.show()
```

```python
print("Skewness:", penguins["body_mass_g"].skew())
```

```python

tips = sns.load_dataset("tips")

sns.countplot(data=tips, x="day")
plt.title("Bills per day")
plt.show()

sns.histplot(data=tips, x="total_bill", kde=True)
plt.title("Total bill distribution")
plt.show()
```

```python
diamonds = sns.load_dataset("diamonds")

sns.countplot(data=diamonds, x="cut",
              order=["Fair", "Good", "Very Good", "Premium", "Ideal"])
plt.title("Diamonds per cut")
plt.show()

sns.histplot(data=diamonds, x="price", kde=True)
plt.title("Diamond price distribution")
plt.show()
```
