```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
sns.boxplot(data=penguins, x="species", y="body_mass_g")
plt.title("Body mass by species")
plt.show()
```

```python
sns.violinplot(data=penguins, x="species", y="body_mass_g")
plt.title("Body mass by species (violin)")
plt.show()
```

```python
sns.boxplot(data=penguins, x="species", y="body_mass_g", hue="sex")
plt.title("Body mass by species and sex")
plt.show()
```

```python
sns.histplot(data=penguins, x="body_mass_g", hue="species", kde=True)
plt.title("Body mass distribution by species")
plt.show()
```

```python
penguins.groupby("species")["body_mass_g"].agg(["mean", "median", "std"]).round(0)
```

```python
tips = sns.load_dataset("tips")

sns.boxplot(data=tips, x="day", y="total_bill", hue="sex")
plt.title("Bill by day and sex")
plt.show()

tips.groupby("day")["total_bill"].agg(["mean", "median", "std"]).round(2)
```

```python
diamonds = sns.load_dataset("diamonds")

sns.boxplot(data=diamonds, x="cut", y="price",
            order=["Fair", "Good", "Very Good", "Premium", "Ideal"])
plt.title("Price by cut")
plt.show()
```

```python
titanic = sns.load_dataset("titanic")

sns.boxplot(data=titanic, x="class", y="fare")
plt.show()
```
